### 4. Graphing random variables

Generate:

- $X$ a random variable, 100 repetitions of a $U(0,1)$ distribution (use $123$ as your seed).
- $Y$ a random variable, 100 repetitions of a normal distribution with zero mean and variance equal to 2, using the same seed.
- Verify the variance of $Y$ with `np.var`.
- Plot $X$ and $Y$ in a scatter plot. Include a title and axis labels in the graph.


```python
import numpy as np #import required packages
from matplotlib.pyplot import subplots

# Generate random number generator and variables
my_rng = np.random.default_rng(123)
X = my_rng.uniform(size=100)
sd = 2**0.5
Y = my_rng.normal(0,sd,100)

# Verify variance (hint: variance gets closer and closer to 2 as you increase the sizes of X and Y to 1000, 10,000, 1000,000...)
print(np.var(Y))

# Plot X and Y
my_fig, my_ax = subplots(figsize=(7,7))
my_ax.plot(X,Y,'s')
my_ax.set_xlabel("X-axis")
my_ax.set_ylabel("Y-axis")
my_ax.set_title("Plot of X vs Y")
```
